## Define Global Variables and Utility Imports

```python
import os
import cv2
import numpy as np
import pandas as pd
from datetime import datetime
from collections import Counter
from sklearn.metrics import classification_report, accuracy_score, precision_score, f1_score
from tensorflow.keras.preprocessing import image

class_names = [
    'Caddisfly',
    'Dipteran',
    'Mayfly',
    'Other',
    'Stonefly',
    'Terrestrial'
]

IMG_SIZE = (224, 224)

HISTORY_CSV = "/content/history.csv"
DIST_CSV = "/content/distribution.csv"
```

## Crop Extraction Function

```python
def extract_crops(dataset, output_dir="/content/new_test_crops"):
    img_dir = os.path.join(dataset.location, "test", "images")
    label_dir = os.path.join(dataset.location, "test", "labels")

    os.makedirs(output_dir, exist_ok=True)

    for cls in class_names:
        os.makedirs(os.path.join(output_dir, cls), exist_ok=True)

    for file in os.listdir(label_dir):
        if not file.endswith(".txt"):
            continue

        img_file = file.replace(".txt", ".jpg")
        img_path = os.path.join(img_dir, img_file)

        if not os.path.exists(img_path):
            continue

        img = cv2.imread(img_path)
        h, w = img.shape[:2]

        with open(os.path.join(label_dir, file)) as f:
            for i, line in enumerate(f):
                cls_id, x, y, bw, bh = map(float, line.split())

                x1 = int((x - bw/2) * w)
                y1 = int((y - bh/2) * h)
                x2 = int((x + bw/2) * w)
                y2 = int((y + bh/2) * h)

                crop = img[y1:y2, x1:x2]

                if crop.size == 0:
                    continue

                cls_name = class_names[int(cls_id)]
                save_path = os.path.join(output_dir, cls_name, f"{file}_{i}.jpg")

                cv2.imwrite(save_path, crop)

    return output_dir
```

```python
crop_folder = extract_crops(new_dataset)
```

## Model Evaluation Function

```python
def evaluate_model(model, crop_folder):
    y_true = []
    y_pred = []

    for cls_idx, cls_name in enumerate(class_names):
        folder = os.path.join(crop_folder, cls_name)

        if not os.path.exists(folder):
            continue

        for file in os.listdir(folder):
            img_path = os.path.join(folder, file)

            img = image.load_img(img_path, target_size=IMG_SIZE)
            arr = image.img_to_array(img) / 255.0
            arr = np.expand_dims(arr, axis=0)

            preds = model.predict(arr, verbose=0)[0]
            pred_idx = np.argmax(preds)

            y_true.append(cls_idx)
            y_pred.append(pred_idx)

    # Fix: Explicitly pass all expected labels to classification_report
    report = classification_report(y_true, y_pred, target_names=class_names, output_dict=True, labels=list(range(len(class_names))), zero_division=0)

    acc = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred, average='weighted', zero_division=0)
    f1 = f1_score(y_true, y_pred, average='weighted', zero_division=0)

    return {
        "accuracy": acc,
        "precision": precision,
        "f1": f1,
        "report": report
    }
```

```python
old_res = evaluate_model(model_cls_old, crop_folder)
new_res = evaluate_model(model_cls_new, crop_folder)
```

## Save Evaluation History

```python
def save_history(model_name, results):
    row = {
        "timestamp": datetime.now(),
        "model": model_name,
        "accuracy": results["accuracy"],
        "precision": results["precision"],
        "f1": results["f1"],
    }

    # add per-class accuracy
    for cls in class_names:
        row[f"{cls}_recall"] = results["report"][cls]["recall"]

    df = pd.DataFrame([row])

    if os.path.exists(HISTORY_CSV):
        df.to_csv(HISTORY_CSV, mode='a', header=False, index=False)
    else:
        df.to_csv(HISTORY_CSV, index=False)
```

```python
save_history("old_model", old_res)
save_history("new_model", new_res)
```

## Compute and Save Class Distribution

```python
def compute_distribution(dataset):
    label_dir = os.path.join(dataset.location, "train", "labels")

    counts = Counter()

    for file in os.listdir(label_dir):
        if not file.endswith(".txt"):
            continue

        with open(os.path.join(label_dir, file)) as f:
            for line in f:
                cls_id = int(line.split()[0])
                counts[cls_id] += 1

    total = sum(counts.values())

    return {
        class_names[i]: counts[i] / total if total > 0 else 0
        for i in range(len(class_names))
    }
```

```python
dist_old = compute_distribution(old_dataset)
dist_new = compute_distribution(new_dataset)
```

```python
def save_distribution(name, dist):
    row = {"dataset": name, "timestamp": datetime.now(), **dist}
    df = pd.DataFrame([row])

    if os.path.exists(DIST_CSV):
        df.to_csv(DIST_CSV, mode='a', header=False, index=False)
    else:
        df.to_csv(DIST_CSV, index=False)
```

```python
save_distribution("old_dataset", dist_old)
save_distribution("new_dataset", dist_new)
```

## Agent Functions for Comparison and Shift Detection

```python
def get_latest_comparison(history):
    old_model = history[history["model"] == "old_model"].iloc[-1]
    new_model = history[history["model"] == "new_model"].iloc[-1]

    return old_model, new_model
```

```python
def get_distribution_shift(distribution):
    old_dist = distribution[distribution["dataset"] == "old_dataset"].iloc[-1]
    new_dist = distribution[distribution["dataset"] == "new_dataset"].iloc[-1]

    shift = {}

    for cls in class_names:
        shift[cls] = new_dist[cls] - old_dist[cls]

    return old_dist, new_dist, shift
```

```python
def structured_agent(old_model, new_model, shift):
    explanation = []

    dip_old = old_model["Dipteran_recall"]
    dip_new = new_model["Dipteran_recall"]

    acc_old = old_model["accuracy"]
    acc_new = new_model["accuracy"]

    # Dipteran priority
    if dip_new > dip_old:
        explanation.append("New model improves Dipteran detection.")
    else:
        explanation.append("Old model performs better on Dipteran.")

    # Overall accuracy
    if acc_new > acc_old:
        explanation.append("New model has higher overall accuracy.")
    else:
        explanation.append("Old model has higher overall accuracy.")

    # Distribution shift
    if abs(shift["Dipteran"]) > 0.1:
        explanation.append("Significant Dipteran distribution shift detected.")

    # Final decision
    if dip_new > dip_old and acc_new >= acc_old:
        decision = "USE NEW MODEL"
    elif dip_old > dip_new:
        decision = "KEEP OLD MODEL"
    else:
        decision = "COLLECT MORE DATA"

    return {
        "decision": decision,
        "explanation": explanation,
        "metrics": {
            "old_accuracy": acc_old,
            "new_accuracy": acc_new,
            "dipteran_old": dip_old,
            "dipteran_new": dip_new
        },
        "shift": shift
    }
```

### LLM Explanation Function

```python
def explain_with_llm(agent_output):
    prompt = f"""
    You are an AI assistant helping ecological researchers.

    Decision: {agent_output['decision']}

    Explanation:
    {agent_output['explanation']}

    Metrics:
    {agent_output['metrics']}

    Distribution shift:
    {agent_output['shift']}

    Explain clearly:
    - Which model should be used
    - Why (focus on Dipteran importance)
    - Whether dataset shift matters
    """

    response = model.generate_content(prompt)
    return response.text
```
